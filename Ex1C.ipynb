```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 15, 7, 20, 12])


plt.figure(figsize=(7, 4))
plt.plot(x, y, color='blue', marker='o', linewidth=2, label='y values') 
plt.title('Line Plot'); plt.xlabel('X'); plt.ylabel('Y')
plt.grid(True); plt.legend(); plt.show()


plt.figure(figsize=(7, 4))
plt.bar(x, y, color='green', edgecolor='black')               
plt.title('Bar Chart'); plt.xlabel('X'); plt.ylabel('Y')
plt.show()


plt.figure(figsize=(7, 4))
plt.scatter(x, y, color='red', s=100, marker='*')             
plt.title('Scatter Plot'); plt.xlabel('X'); plt.ylabel('Y')
plt.grid(True); plt.show()


plt.figure(figsize=(6, 6))
labels = ['A', 'B', 'C', 'D', 'E']
plt.pie(y, labels=labels, autopct='%1.1f%%', startangle=90)   
plt.title('Pie Chart'); plt.show()
```
