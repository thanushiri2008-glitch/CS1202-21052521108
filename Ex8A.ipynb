## AIM

To perform Time Series Analysis on monthly data using Moving Average and ARIMA models for forecasting future values.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas, numpy, matplotlib and ARIMA from statsmodels.tsa.arima.model.

**Step 3:** Create a monthly airline passengers DataFrame with 48 months of data.

**Step 4:** Set the Date column as the index of the DataFrame and set its frequency to monthly (MS).

**Step 5:** Plot the original time series.

**Step 6:** Compute the 12-month rolling mean (Moving Average).

**Step 7:** Plot the original series along with the Moving Average line.

**Step 8:** Fit an ARIMA(1, 1, 1) model on the Passengers column using statsmodels.

**Step 9:** Display the summary of the fitted ARIMA model.

**Step 10:** Forecast the next 12 months and plot the forecast with historical data.

**Step 11:** Stop the program.



## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA

dates = pd.date_range(start='2020-01-01', periods=48, freq='MS')
passengers = [112,118,132,129,121,135,148,148,136,119,104,118,
              115,126,141,135,125,149,170,170,158,133,114,140,
              145,150,178,163,172,178,199,199,184,162,146,166,
              171,180,193,181,183,218,230,242,209,191,172,194]
df = pd.DataFrame({'Date': dates, 'Passengers': passengers})
df.set_index('Date', inplace=True)
df.index.freq = 'MS'                                      # MS = month start frequency
print("First 5 rows:")
print(df.head())

plt.figure(figsize=(10, 5))
plt.plot(df.index, df['Passengers'], color='blue', marker='o', label='Original')
plt.title('Monthly Airline Passengers')
plt.xlabel('Date'); plt.ylabel('Passengers')
plt.grid(True); plt.legend(); plt.show()

df['MA_12'] = df['Passengers'].rolling(window=12).mean()  # 12-month moving average
plt.figure(figsize=(10, 5))
plt.plot(df.index, df['Passengers'], color='blue', marker='o', label='Original')
plt.plot(df.index, df['MA_12'], color='red', linewidth=2.5, label='12-Month MA')
plt.title('Original vs Moving Average')
plt.grid(True); plt.legend(); plt.show()

model  = ARIMA(df['Passengers'], order=(1, 1, 1))         # ARIMA(p, d, q) model
fitted = model.fit()
print(fitted.summary())

forecast = fitted.forecast(steps=12)                      # predict next 12 months
forecast_index = pd.date_range(start=df.index[-1] + pd.DateOffset(months=1),
                               periods=12, freq='MS')
plt.figure(figsize=(10, 5))
plt.plot(df.index, df['Passengers'], color='blue', marker='o', label='Historical')
plt.plot(forecast_index, forecast, color='green', marker='s', linewidth=2.5, label='Forecast')
plt.title('ARIMA Forecast - Airline Passengers')
plt.grid(True); plt.legend(); plt.show()

print("Forecasted values (next 12 months):")
for d, v in zip(forecast_index, forecast):
    print(f"  {d.strftime('%Y-%m')} : {v:.2f}")
```
