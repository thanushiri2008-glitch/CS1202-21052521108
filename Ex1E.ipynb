```python
# Ex.No. 1(e)
## PANDAS DATAFRAME OPERATIONS 


AIM

To perform Pandas DataFrame operations — creating a DataFrame from a list of dictionaries with row labels, adding a row (Divya with 95 marks) and a column, slicing rows, deleting a row by its index label, 
and displaying only the Iris-virginica records of the Iris data set.

ALGORITHM

Step 1: Start the program.

Step 2: Import the pandas package as pd.

Step 3: Create a DataFrame from a list of dictionaries with the row labels s1, s2 and s3.

Step 4: Add a new row [‘Divya’, 95] using loc and a new column Bonus (Marks + 5), so that Divya’s Bonus is 100.

Step 5: Slice rows 2 to 3 using df[1:3].

Step 6: Delete the row s2 using drop().

Step 7: Read the Iris data set from the CSV file using read_csv().

Step 8: Display only the Iris-virginica records using a condition on the species column.

Step 9: Stop the program.
```

```python
import os
import pandas as pd

# ---- Setup: create data/iris.csv only if it does not exist ----
os.makedirs('data', exist_ok=True)
if not os.path.exists('data/iris.csv'):
    cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
    try:
        from sklearn.datasets import load_iris
        d = load_iris(as_frame=True).frame
        d.columns = cols + ['target']
        d['species'] = d['target'].map({0: 'Iris-setosa',
                                        1: 'Iris-versicolor',
                                        2: 'Iris-virginica'})
        d = d.drop(columns='target')
    except ImportError:                      # if scikit-learn is not installed
        rows = [[5.1, 3.5, 1.4, 0.2, 'Iris-setosa'],
                [4.9, 3.0, 1.4, 0.2, 'Iris-setosa'],
                [4.7, 3.2, 1.3, 0.2, 'Iris-setosa'],
                [7.0, 3.2, 4.7, 1.4, 'Iris-versicolor'],
                [6.4, 3.2, 4.5, 1.5, 'Iris-versicolor'],
                [6.9, 3.1, 4.9, 1.5, 'Iris-versicolor'],
                [6.3, 3.3, 6.0, 2.5, 'Iris-virginica'],
                [5.8, 2.7, 5.1, 1.9, 'Iris-virginica'],
                [7.1, 3.0, 5.9, 2.1, 'Iris-virginica'],
                [6.3, 2.9, 5.6, 1.8, 'Iris-virginica'],
                [6.5, 3.0, 5.8, 2.2, 'Iris-virginica']]
        d = pd.DataFrame(rows, columns=cols + ['species'])
    d.to_csv('data/iris.csv', index=False)

# ---- Main program ----
# DataFrame from a list of dictionaries, with row labels
data = [{'Name': 'Arun', 'Marks': 85},
        {'Name': 'Bala', 'Marks': 72},
        {'Name': 'Chitra', 'Marks': 90}]
df = pd.DataFrame(data, index=['s1', 's2', 's3'])     # index = row labels
print(df)

# Append (add) a new row and a new column
df.loc['s4'] = ['Divya', 95]                          # Task 1: changed 78 to 95
df['Bonus'] = df['Marks'] + 5                         # new column from an old column
print("After adding a row and a column:")
print(df)

# Slicing rows
print("Rows 2 to 3:")
print(df[1:3])                                        # rows 1 and 2 (end is not included)

# Delete a row by its index label
df = df.drop('s2')                                    # delete the row with label s2
print("After deleting s2:")
print(df)

# Iris: load the CSV file and display only Iris-virginica
iris = pd.read_csv('data/iris.csv')
print(iris[iris['species'] == 'Iris-virginica'].head())  # Task 2: changed 'Iris-setosa' to 'Iris-virginica'
```
